# NUMBER GUESSING GAME
    


```python
import random
secret_number=random.randint(1,10)

attempt=3
while attempt>0:
    guess_number=int(input('guess a number between 1 and 10: '))
    attempt-=1
    if guess_number>10:
        print('number is out of the valid range')
        continue
    if guess_number>secret_number:
        print('number you guess is greaterthan secret number')
    elif guess_number<secret_number:
        print('number you guess is lowerthan secret number')
    else:
        print("congratulations")
        print()
        print('your guess is correct')
        break
else:
    print('better luck next time')
```

# MULTIPLICATION TABLE GENERATOR

```python
number=int(input('Enter the number for which you want the multiplication table: '))
for i in range(1,11):
    print(number,"x",i,"=",number*i)
    i+=1
```

# BMI CALCULATOR

```python
def calculate_bmi(weight, height):
    print('Your BMI is: ',weight/((height)**2))
wei=float(input('Enter your weight in kg: '))
hei=float(input('Enter your height in meters: '))
calculate_bmi(wei, hei)
```
